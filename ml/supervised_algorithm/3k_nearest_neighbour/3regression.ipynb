```python
import pandas as pd


df=pd.read_csv("../house.csv")
```

```python
X = df.drop(columns=["price"])
y=df["price"]
#splitting data in train test
from sklearn.model_selection  import train_test_split
X_train, X_test, y_train,y_test=train_test_split(X,y, test_size=0.2, random_state=2)
y_train
```

```python
#making pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
#categorical columns  and numerical columns
df_categorical_columns = X.select_dtypes(include=["object"]).columns
df_numerical_columns = X.select_dtypes(include=["int64","float64"]).columns
df_categorical_columns
```

```python
preprocessor = ColumnTransformer([
    ("num",StandardScaler(), df_numerical_columns), #if you dont do standardScaler you can see r2 score will highly drop
    ("cat",OneHotEncoder(),df_categorical_columns )
])
```

```python
from sklearn.neighbors import  KNeighborsRegressor


pipeline = Pipeline([
    ("preprocessor",preprocessor), #it takes input features
    # ("model",LinearRegression()) #while training it takes output of preprocessor and output, while testing it take only input of preprocesor
    # ("model",LinearRegression()) #while training it takes output of preprocessor and output, while testing it take only input of preprocesor
    ("model",KNeighborsRegressor()) #while training it takes output of preprocessor and output, while testing it take only input of preprocesor
])
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "model__n_neighbors": [3, 5, 7, 9, 11],
}

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="r2",
   
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Test R2:", grid_search.score(X_test, y_test))
```
