```python
import pandas as pd
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler

# Training data
df_train = pd.DataFrame({
    "year": [2073, 2074, 2075, 2076, 2077, 2078, 2079, 2080, 2081, 2082, 2083],
    "gold_price": [
        53000, 58000, 70000, 85000, 99800,
        120000, 150000, 180000, 250000, 286000, 316700
    ]
})

# Test data
df_test = pd.DataFrame({
    "year": [2084, 2085, 2086, 2087, 2088]
})

# -------------------------
# 1. Separate X and y
# -------------------------

X_train = df_train.drop(columns=["gold_price"])
y_train = df_train["gold_price"]

X_test = df_test

# -------------------------
# 2. Scale X
# -------------------------

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# -------------------------
# 3. Create SGD model
# -------------------------

model = SGDRegressor(
    learning_rate="constant",
    eta0=0.01,
    max_iter=10000,
    tol=1e-6,
    random_state=42
)

# -------------------------
# 4. Train
# -------------------------

model.fit(X_train_scaled, y_train)

# -------------------------
# 5. Predict
# -------------------------

predictions = model.predict(X_test_scaled)

# -------------------------
# 6. Display predictions
# -------------------------

df_test["predicted_gold_price"] = predictions.round(2)

print(df_test)
```
