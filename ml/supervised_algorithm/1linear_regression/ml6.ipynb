#methode to prevent from overfitting
## Generally overfit model has Coefficients(slope coeffient not constant) unnecessarily large, it means small change in input feature may result high prediction
## lasso and rigre regression  Both add a penalty for large coefficients:
## lasso and ridge regression shrink coefficient to reduce overfitting
##lasso => l1 regularization it add   alpha*|sum of coeffiecient| in cost funtion(mean square error) ie mse+alpha * sum of slope coeffiecint
##ridge => l2 regularization it add  alpha*(sum of square of coefficient) it is named l2 because it is square(2)

## (intercept/constant) is normally not included in the Lasso penalty.

Higher alpha→stronger penalty→more coefficient shrinking
if you increase to much alpha => slop will be very less it means output is less depend on input it result underfit
	​


and

Lower alpha→weaker penalty→less coefficient shrinking
	​



```python
df=pd.read_csv("./dataset/house.csv")
```

```python
df
```

```python
#checking null value
df.isnull().sum().sum()
```

```python

X = df.drop(columns=["price"])
y=df["price"]
```

```python

X = df.drop(columns=["price"])
y=df["price"]
#splitting data in train test
from sklearn.model_selection  import train_test_split
X_train, X_test, y_train,y_test=train_test_split(X,y, test_size=0.2, random_state=2)
y_train
```

```python
#making pipeline
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
```

```python
#categorical columns  and numerical columns
df_categorical_columns = X.select_dtypes(include=["object"]).columns
df_numerical_columns = X.select_dtypes(include=["int64","float64"]).columns
df_categorical_columns
```

```python
preprocessor = ColumnTransformer([
    ("num",StandardScaler(), df_numerical_columns), #if you dont do standardScaler you can see r2 score will highly drop
    ("cat",OneHotEncoder(),df_categorical_columns )
])
```

```python
#creating pipelin
from sklearn.linear_model import Lasso,Ridge
pipeline = Pipeline([
    ("preprocessor",preprocessor), #it takes input features
   ("model",LinearRegression()) 
    # ("model", Lasso(alpha=1))#you can tune to alpha to what ever value
      # ("model",Ridge(alpha=1))#you can tune alpha to what ever value
])
```

```python
#training the model
pipeline.fit(X_train,y_train)
```

```python
#r2 score
print("training accuracy", pipeline.score(X_train,y_train))
print("testing accuracy", pipeline.score(X_test,y_test))
# print("intercept", pipeline.named_steps["model"].intercept_)
# print("coefficients", pipeline.named_steps["model"].coef_)
```

```python
#saving the model using joblib package
#using joblib we can save
#model, dataframe
#in google colab it is recommend that once you dump downoad it and store in local computer, in another session , data will be deleted automateically
import joblib
joblib.dump(pipeline,"house_price_prediction.joblib")#or use .pkl
joblib.dump(pipeline,"house_price_prediction.pkl")#or use .pkl
```
