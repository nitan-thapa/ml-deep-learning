```python
#class_weight
# it is not sampling process
#Give more importance to the minority class and less importance to the majority class.
#ie the algorithm try be improve the performance of the minority class or say reduce error in the minority class

# or
# class_weight="balanced"
# → does not change the data
# → changes the importance/penalty of classes during training
```

```python
import pandas as pd


df = pd.read_csv('imbalanced_data.csv')
df.head()
df["Class"].value_counts()
```

```python
X =df.drop(columns=["Class"])
y=df["Class"]
```

```python
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report


m1 = LogisticRegression()
m1.fit(X_train, y_train)
y1_pred = m1.predict(X_test)
print(classification_report(y_test, y1_pred))
```

```python
#after applyng class weight

m1 = LogisticRegression(class_weight="balanced")
m1.fit(X_train, y_train)
y1_pred = m1.predict(X_test)
print(classification_report(y_test, y1_pred))
```
