```python
#SMOTE
#synthetic minority over-sampling technique
#it increase the number of minority class 
#it doesn't take duplcate values but it takes near values

#or
# It creates new synthetic samples
# based on existing minority-class samples
# and their nearest neighbors.

```

```python
import pandas as pd


df = pd.read_csv('imbalanced_data.csv')
df.head()
df["Class"].value_counts()
```

```python
X =df.drop(columns=["Class"])
y=df["Class"]
```

```python
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report


m1 = LogisticRegression()
m1.fit(X_train, y_train)
y1_pred = m1.predict(X_test)
print(classification_report(y_test, y1_pred))
```

```python
#applying random oversampling on training data 
# pip install imbalanced-learn
from imblearn.over_sampling import SMOTE  # synthetic minority over-sampling technique

ros = SMOTE(random_state=42)
X_resampled, y_resampled = ros.fit_resample(X_train, y_train)


print(y_resampled.value_counts())

```

```python
m2 = LogisticRegression()
m2.fit(X_resampled, y_resampled)
y2_pred = m2.predict(X_test)
print(classification_report(y_test, y2_pred))

#here we increase recall for 1 but its precision is reduced 
#but we implement undersampling on training data so our model perform well on the minority class
```
