```python
#imbalace data are the data where the target variable is imbalanced
#if we train the model on the imbalanced data then the model will perform well on the majority class but will not perform well on the minority class
#to solve this problem we use undersampling
#undersampling is the process of reducing the number of samples in the maority class (sample are removed randomly)
#undersampling is done by removing some of the samples in the majority class

#note sample is only done in the training data not in the test data

#advantage 
#can perform well on the minority class
#training time is reduced

#disadvantage
#we lose some of the information from the majority class


```

```python
import pandas as pd


df = pd.read_csv('imbalanced_data.csv')
df.head()
df["Class"].value_counts()
```

```python
X =df.drop(columns=["Class"])
y=df["Class"]
```

```python
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report


m1 = LogisticRegression()
m1.fit(X_train, y_train)
y1_pred = m1.predict(X_test)
print(classification_report(y_test, y1_pred))
```

```python
#applying random undersampling on training data 
# pip install imbalanced-learn
from imblearn.under_sampling import RandomUnderSampler
rus = RandomUnderSampler(random_state=42)#1:1 ratio
X_resampled, y_resampled = rus.fit_resample(X_train, y_train)


print(y_resampled.value_counts())

```

```python
m2 = LogisticRegression()
m2.fit(X_resampled, y_resampled)
y2_pred = m2.predict(X_test)
print(classification_report(y_test, y2_pred))

#here we increase recall for 1 but its precision is reduced 
#but we implement undersampling on training data so our model perform well on the minority class
```
