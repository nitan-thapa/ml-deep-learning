Average (mean, median, mode)

Numerical 

    mean  (it is not used if data has outlier)

    median  (it is used if data has outlier)

Categorical 

    mode  (it is used for categorical variable)



Outlier 

    it can only effect mean

Q1(25%)  => 25% of data lies below q1 value

Q2(50%)  => 50% of data lies below q2 value

Q3(75%)  => 75% of data lies below q3 value
