```python
import matplotlib.pyplot as plt
```

```python
#line plot
#plot(x,y)
#xlabel, ylabel, title, show
day=["sun","mon", "tue","wed","thu","fri","sat"]
share_price=[200,205,210,210,200,225,205]
plt.plot(day,share_price)
plt.title("Upper")
plt.xlabel("day")
plt.ylabel("Share price in Rs")
plt.show()
```

```python
#custumization plot
#grid

plt.plot(day,share_price, marker="o", color="red", linestyle="dashed", linewidth=4, markersize=10,label="sales data")
plt.title("Upper")
plt.xlabel("day")
plt.ylabel("Share price in Rs")
# plt.grid()
plt.grid(color="grey", linestyle="--", linewidth=0.2)
plt.show()
```

```python
#xticks
#converting number x-tick to string x-tick
#rotation x-tick
day=[1,2,3,4,5,6,7]
share_price=[200,205,210,210,200,225,205]
plt.plot(day,share_price)
plt.title("Upper")
plt.xlabel("week")
plt.ylabel("Share price in Rs")
plt.xticks(day,["sunday","monday","tuesday","wednesday","thrusday","firday", "saturday"])
plt.xticks(rotation=45)
plt.show()
```
