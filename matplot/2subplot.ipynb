```python
#figure 
#subplot
#tight_layout


day=["sun","mon", "tue","wed","thu","fri","sat"]
share_price_ntc=[200,205,210,210,200,225,205]
share_price_ncell= [180,185,190,190,180,205,185]
share_price_nric = [160,165,170,170,160,185,165]
share_price_hrl = [140,145,150,150,140,165,145]
share_price_umhl = [120,125,130,130,120,145,125]
share_price_kbl = [100,105,110,110,100,125,105]



```

```python
from matplotlib import pyplot as plt

#plt.subplot(row, column, position)

plt.figure(figsize=(10,5))# here whole figure size is width=10inch, height as 5inch

""" 
1, 2, 3, 4
5, 6

 """
# here note 1 figure only takes space as as per suplot size

plt.subplot(2,4,1)#we create  2 rows, 3 col and place the plot at 1 position
plt.plot(day,share_price_ntc)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("NTC")
plt.xticks(rotation=45)



plt.subplot(2,4,2)
plt.plot(day,share_price_ncell)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("NCELLC")
plt.xticks(rotation=45)

plt.subplot(2,4,3)
plt.plot(day,share_price_nric)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("NRIC")
plt.xticks(rotation=45)

plt.subplot(2,4,4)
plt.plot(day, share_price_hrl)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("HRL")
plt.xticks(rotation=45)


plt.subplot(2,4,5)
plt.plot(day,share_price_umhl)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("UMHL")
plt.xticks(rotation=45)

plt.subplot(2,4,6)
plt.plot(day,share_price_kbl)
plt.xlabel("Day")
plt.ylabel("Share Price")
plt.title("KBL")
plt.xticks(rotation=45)


plt.tight_layout()#it automaitcally adjuest spacing between subplot



plt.show()

```
