```python
import numpy as np
import pandas as pd


train_data = pd.DataFrame({
    "Age": [20, 21, 22, 20, 23, 21, 22, np.nan, 19, 25],
    
    "Study_Hours": [
        5, 6, 4, np.nan, 7,
        5, 6, 4, 10, 5
    ],
    
    "Gender": [
        "Male", "Female", "Male", "Female", "Male",
        "Female", np.nan, "Male", "Female", "Male"
    ],

    
})

print(train_data)

#  numerical column   => Age, study hour  =   simpleimputer(strategy="mean")  =>  meanmaxscaler()
#  categorical column  => gender  =   simpleimputer(strategy="most_frequent")  =>  onehotencoder
```

```python
numerical_feature =  train_data.select_dtypes(include=["int64", "float64"]).columns#["Age","Study_Hours"]
categorical_feature = train_data.select_dtypes(include=["str"]).columns#["Gender"]
```

```python


from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder

numerical_pipeline= Pipeline(#first simpleImputer will work then min  min max scaaler will work
    [
        ("imputer", SimpleImputer(strategy="mean")),
        ("minmaxscaling", MinMaxScaler())
    ]
)

categorical_pipeline = Pipeline(
    [
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("ohe", OneHotEncoder())
    ]
)



preprocessor=   ColumnTransformer(#
    [
        ("numercial", numerical_pipeline, numerical_feature),#["Age","Study_Hours"]
        ("categorical", categorical_pipeline, categorical_feature)#["Gender"]
    ],
    verbose_feature_names_out=False
)
```

```python
all_featrues  =  preprocessor.fit_transform(train_data)
train_data[preprocessor.get_feature_names_out()] = all_featrues
train_data
```

```python
test_data = pd.DataFrame({
    "Age": [22, np.nan, 24],
    
    "Study_Hours": [6, 3, np.nan],
    
    "Gender": ["Female", np.nan, "Female"],
    
})

print(test_data)
```

```python
test_featrues = preprocessor.transform(test_data)
test_data[preprocessor.get_feature_names_out()] = test_featrues
test_data
```
