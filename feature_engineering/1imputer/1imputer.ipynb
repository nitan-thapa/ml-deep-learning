```python
#before diving first learn train_test_split.png image  

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [
        "Ram", "Shyam", "Hari", "Sita", "Gita",
        "Nabin", "Rita", "Bikash", "Anita", "Suman"
    ],

    # Categorical variables
    "Gender": [
        "Male", "Male", np.nan, "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],  # 10% missing

    "City": [
        "Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur", "Kathmandu",
        "Pokhara", "Lalitpur", "Kathmandu", "Pokhara", np.nan
    ],  # 10% missing

    "Department": [
        "IT", np.nan, np.nan, "Finance", "IT",
        np.nan, "HR", np.nan, "IT", "Finance"
    ],  # 40% missing

    # Numerical variables
    "Age": [
        20, 22, np.nan, 21, 23,
        24, 25, 26, 22, 21
    ],  # 10% missing

    "Math": [
        80, 75, 90, np.nan, 85,
        78, 88, 92, 70, 95
    ],  # 10% missing

    "Salary": [
        30000, np.nan, np.nan, 40000, np.nan,
        35000, np.nan, 50000, np.nan, 38000
    ]  # 50% missing 
})

df
```

```python

#fill numerical colum by mean useing SimpleImputer()
df_numeric = df.select_dtypes(include=["int64", "float64"])
```

```python
#impute means filling 
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy ="mean") #medain, most_frequent
imputerValue=imputer.fit_transform(df_numeric)#it learn mean for numerical colums of df
print(imputer.get_feature_names_out())
df[imputer.get_feature_names_out()] =imputerValue
df
```

```python
#once you made fit pipleline you can use that pipeline for any data set (mostly used for test dataset)
#usually we learn mean...from train dataset and apply the same mean to test dataset 

#if our test dataset is 
test_df = pd.DataFrame({
    "Name": [
        "Aashish", "Kiran"
    ],

    "Gender": [
        "Male", np.nan
    ],

    "City": [
        "Pokhara", "Kathmandu"
    ],

    "Department": [
        np.nan, "Finance"
    ],

    "Age": [
        23, np.nan
    ],

    "Math": [
        88, 76
    ],

    "Salary": [
        np.nan, 42000
    ]
})

```

```python
test_df[imputer.get_feature_names_out()]=imputer.transform(test_df[imputer.get_feature_names_out()])
```
