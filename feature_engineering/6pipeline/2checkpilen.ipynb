```python
import pandas as pd
import numpy as np
from  sklearn.pipeline  import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

train_data = pd.DataFrame({
    "Age": [20, 21, 22, 20, 23, 21, 22, 24, 19, 25],
    
    "Study_Hours": [
        5, 6, 4, np.nan, 7,
        5, 6, 4, 10, 5
    ],
    
    "Gender": [
        "Male", "Female", "Male", "Female", "Male",
        "Female", np.nan, "Male", "Female", "Male"
    ],
    "Result":[1,0,1,0,1,0,1,1,0,1]
    
})

print(train_data)
```

```python
test_data = pd.DataFrame({
    "Age": [22, np.nan, 24],
    
    "Study_Hours": [6, 3, np.nan],
    
    "Gender": ["Female", "Male", "Female"],
    "Result":[1,0,1]
    
})

print(test_data)
```

```python
# make pipeline form train data and  implement that pipeline on a dest dat 
# fill numerical value by median
# fill categorical value by most_frequent
#encode categorical value by one hot encoding
```

```python
num_col = train_data.select_dtypes(include=["int64", "float64"]).columns
cat_col = train_data.select_dtypes(include=["str"]).columns
```

```python
#numerical pipeline 

num_pipeline = Pipeline(
    [
        ("imputer", SimpleImputer(strategy="mean"))

    ]
)

```

```python
#categorical pipeline 
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(drop="first"))
])
```

```python
#combine pipelein 

pipeline = ColumnTransformer(
    [
        ("numerical", num_pipeline,num_col),
        ("categorical", cat_pipeline,cat_col)
    ],
     verbose_feature_names_out=False # it does not change column name
)
```

```python
pipeline.fit_transform(train_data)  #here note though it congain Result which is not part of ColumnTransformer , it will work
# pipeline.fit_transform(train_data[["Age", "Study_Hours", "Gender"]])
# pipeline.fit_transform(train_data[["Gender", "Age", "Study_Hours"]])#we can chage column order
# pipeline.fit_transform(train_data[["Age","Gender" ]])# error we can chage column order but all column used by ColumnTransforme must be present
# pipeline.fit_transform(train_data[["Age","Gender" ]])# error we can chage column order but all column used by ColumnTransforme must be present
```

```python
print(pipeline.get_feature_names_out())
```

```python
train_data[pipeline.get_feature_names_out()]  =pipeline.fit_transform(train_data)
```
