```python
#it does not need any inbuilt library
#Categorical data
    #nominal (no mathmetical value,no order)
        #use oneHotEncoder  for input variable
        #use labelEncoder for output variable
    #ordinal (has mathmetical value,order)
        #use  ordinalEncoder for input variable
        #use ordinalEncoder for output variable

import pandas as pd

df = pd.DataFrame({
    "Study Hours": [2, 3, 4, 5, 6, 7, 1, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 50, 90],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female"],
    "Result": ["Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Fail", "Pass"]
})

print(df)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df["Result"] = encoder.fit_transform(df["Result"])#it convert  result class to 0 and 1 ,it can treate 0 to any and 1 to any
df

```

```python
#but if you want to convert manually use map 

df["Result"] = df["Result"].map({"Pass":1,"Fail":0})#convert manually. (a.map() ony words with pandas)
df
```
