```python
#Categorical data
    #nominal (no mathmetical value,no order)
        #use oneHotEncoder  for input variable
        #use labelEncoder for output variable
    #ordinal (has mathmetical value,order)
        #use  ordinalEncoder for input variable
        #use ordinalEncoder for output variable
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 25, 28, 30, 35, 40, 27, 45],
    "Experience": [1, 2, 4, 5, 8, 12, 3, 15],
    "Education": [
        "Bachelor", "Bachelor", "Master", "Master",
        "Bachelor", "Master", "Bachelor", "Master"
    ],
    "Performance Level": [
        "Poor", "Average", "Good", "Excellent",
        "Good", "Excellent", "Average", "Excellent"
    ]
})

print(df)
```

```python


from sklearn.preprocessing import OrdinalEncoder

ordinal_columns = ["Education", "Performance Level"]
#convert class to number  0, 1,2 ..

semantic_order = [
    ["Bachelor", "Master"], #treat bacherlor as 0 , master as 1
    ["Poor", "Average", "Good", "Excellent"] #treat poor as 0 , average as 1...
]

# encoder = OrdinalEncoder.OrdinalEncoder()  #without using symantic order
encoder = OrdinalEncoder(categories=semantic_order)
encoded = encoder.fit_transform(df[ordinal_columns])
df[encoder.get_feature_names_out()] = encoded #no need to  .array()

print(df)

```
