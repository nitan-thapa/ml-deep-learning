```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [
        "Ram", "Shyam", "Hari", "Sita", "Gita",
        "Nabin", "Rita", "Bikash", "Anita", "Suman"
    ],

    # Categorical variables
    "Gender": [
        "Male", "Male", np.nan, "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],  # 10% missing

    "City": [
        "Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur", "Kathmandu",
        "Pokhara", "Lalitpur", "Kathmandu", "Pokhara", np.nan
    ],  # 10% missing

    "Department": [
        "IT", np.nan, np.nan, "Finance", "IT",
        np.nan, "HR", np.nan, "IT", "Finance"
    ],  # 40% missing

    # Numerical variables
    "Age": [
        20, 22, np.nan, 21, 23,
        24, 25, 26, 22, 21
    ],  # 10% missing

    "Math": [
        80, 75, 90, np.nan, 85,
        78, 88, 92, 70, 95
    ],  # 10% missing

    "Salary": [
        30000, 33000, 33000, 40000, 34000,
        35000, 33000, 50000, 35000, 38000
    ]  # 50% missing 
})
```

```python
# preprocess train data
numerical_cols =df.select_dtypes(include=["int64", "float64"]).columns#["Age", "Math", "Salary"]
categorical_cols =df.select_dtypes(include=["str"]).columns#["Name","Gender", "City", "Department"]
```

```python
# preprocessor = ColumnTransformer([("numerical", num_pipeline, numerical_cols), ("categorical", cat_pipeline, categorical_cols)])#hrer order is not importna
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

# For different columns use differen  name
preprocessor = ColumnTransformer(  # ["A","B","C","D","E"]
    [
        ("numerical", SimpleImputer(strategy="mean"),numerical_cols),# ["A"]
        ("categorical",SimpleImputer(strategy="most_frequent"),categorical_cols) #["B","c"]
    ],
      verbose_feature_names_out=False # it does not change features name
)# ["A","B","C","D","E"]
print()
```

```python
value=preprocessor.fit_transform(df)
df[preprocessor.get_feature_names_out()] = value
df
# print(preprocessor.get_feature_names_out())
# df[]
# df[preprocessor.get_feature_names_out()] = value
# value =preprocessor.fit_transform(df)
# df[preprocessor.get_feature_names_out()] = preprocessor.fit_transform(X_train)
# print(preprocessor.get_feature_names_out())
# df
```
