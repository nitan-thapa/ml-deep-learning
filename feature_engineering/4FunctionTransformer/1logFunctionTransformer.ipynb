```python
import pandas as pd
import numpy as np
import seaborn as sns

df = pd.DataFrame({
    "Area": [500, 700, 900, 1200, 1500, 2000, 3000, 5000],
    "Bedrooms": [1, 2, 2, 3, 3, 4, 5, 6],
    "Price": [30000, 40000, 50000, 70000, 90000, 130000, 250000, 600000]
})

print(df)
```

```python
#before log transform 
sns.histplot(df["Price"],kde=True)
#log transform compress the expre value
#it is used to if a distribution is righet skewed
```

```python
from sklearn.preprocessing import FunctionTransformer

log_transfomer = FunctionTransformer(np.log1p)#it means log(1+x) one is added if x =0 it gives invalid  ouput so we add 1


df[["Price"]] = log_transfomer.fit_transform(df[["Price"]])

print(df)
```

```python
sns.histplot(df["Price"],kde=True)
```
