```python
import pandas as pd
import numpy as np
import seaborn as sns

df = pd.DataFrame({
    "Name": [
        "Ram", "Shyam", "Hari", "Sita", "Gita",
        "Nabin", "Rita", "Bikash", "Anita", "Suman",
        "Roshan", "Puja", "Aashish", "Kiran", "Prabin"
    ],

    "Exam_Score": [
        20, 50, 70, 80, 85,
        88, 90, 92, 94, 95,
        96, 97, 98, 99, 100
    ]
})

print(df)
```

```python
sns.histplot(df["Exam_Score"], kde=True)
```

```python
from sklearn.preprocessing import FunctionTransformer

square_transfomer = FunctionTransformer(np.square)
df["Square_Score"] = square_transfomer.fit_transform(df[["Exam_Score"]])
df
```

```python
sns.histplot(df["Square_Score"], kde=True)
```
