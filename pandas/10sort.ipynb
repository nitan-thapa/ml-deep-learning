```python
import pandas as pd
data = {
    "Name": ['nitan', 'ram', 'hari',"gita"],
    "Age": [28, 28, 22,34],
    "Salary": [30000, 20000, 30000,10000]
}
df=pd.DataFrame(data)
```

```python
df.sort_values(by=["Age"],ascending=[True],inplace=False)
```

```python
df.sort_values(by=["Age","Salary"],ascending=[True,False],inplace=False)
```

```python
df.sort_values(by=["Name","Salary"],ascending=[True,False],inplace=False)
```
