```python
import pandas as pd

df = pd.DataFrame({
    "Age": ["20", "21", "22", "23"],
    "Height": ["5.5", "5.7", "5.9", "6.0"],
    "id": [101, 102, 103, 104],
    "Marks": ["80", "90", "abc", "70"]
})
print(df.dtypes)#give types of each column
```

```python
df["Age"] = df["Age"].astype(int)
df["Age"].dtype  #here for single colume use .dtype
```

```python
df["Height"]= df["Height"].astype(float)
df["Height"].dtype
```

```python
df["id"]= df["id"].astype(str)
df["id"].dtype
```

```python
# df["Marks"]= df["Marks"].astype(int) #it will throw an error as abc is not a number
df["Marks"]= pd.to_numeric(df["Marks"], errors="coerce")#if there is any error then it will convert it to NaN  (note NaN is float)
df["Marks"]= pd.to_numeric(df["Marks"], errors="coerce").astype("Int64")#int can not convert NaN to <NA> so int64 is written 
df
```
