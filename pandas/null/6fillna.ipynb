```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [
        "Ram", "Shyam", "Hari", "Sita", "Gita",
        "Nabin", "Rita", "Bikash", "Anita", "Suman",
        "Roshan", "Puja", "Aashish", "Kiran", "Prabin",
        "Sarita", "Dipak", "Manisha", "Ramesh", "Nisha"
    ],

    "Developer": [
        "Frontend", "Backend", "Database", "ML", "Frontend",
        "Backend", "ML", "Database", "Frontend", "Backend",
        "ML", "Database", "Frontend", "ML", "Backend",
        "Database", "ML", "Frontend", "Backend", "Database"
    ],

    "Salary": [
        55000, np.nan, 75000, 90000, np.nan,
        72000, 95000, np.nan, 52000, 68000,
        np.nan, 78000, 65000, np.nan, 71000,
        83000, 88000, np.nan, 69000, 85000
    ]
})

```

```python
df["class_mean_salary"]= df.groupby("Developer")["Salary"].transform("mean")
#it will create new column for each row of developer and it will calculate store  salary for each developer 
print(df)
```

```python
df["Salary"]= df["Salary"].fillna(df["class_mean_salary"])
#here it will replace NaN with value of class_mean_salary
```
