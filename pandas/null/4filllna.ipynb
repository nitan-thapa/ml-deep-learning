```python
#we generall fill column rather than row
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ram", "Shyam", "Hari", "Sita", "Gita"],
    "Age": [20, np.nan, 22, None, 21],
    "Math": [80, 75, np.nan, 90, None],
    "Science": [85, np.nan, 78, 92, 88],
    "City":["Kathmandu","Pokhara","Lalitpur",None,"Lalitpur"]
})

print(df)
```

```python
# df.fillna(value)#it fill all non value with value
df.fillna(0)

```

```python
# df.fillna({"Math":0,"Science":0},inplace=True) #inplace change orginal data
df.fillna({"Math":0,"Science":0})
#it fill Math nan value with 0 and Science nan value with 0 other remains unchanged
```

```python
df.fillna({
    "Math":df["Math"].mean(),
    "Science":df["Science"].mean(),
    "City":df["City"].mode()[0]
     

    },)
```
