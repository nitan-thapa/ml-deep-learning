```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": [
        "Ram", "Shyam", "Hari", "Sita", "Gita",
        "Nabin", "Rita", "Bikash", "Anita", "Suman"
    ],

    # Categorical variables
    "Gender": [
        "Male", "Male", np.nan, "Female", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ],  # 10% missing

    "City": [
        "Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur", "Kathmandu",
        "Pokhara", "Lalitpur", "Kathmandu", "Pokhara", np.nan
    ],  # 10% missing

    "Department": [
        "IT", np.nan, np.nan, "Finance", "IT",
        np.nan, "HR", np.nan, "IT", "Finance"
    ],  # 40% missing

    # Numerical variables
    "Age": [
        20, 22, np.nan, 21, 23,
        24, 25, 26, 22, 21
    ],  # 10% missing

    "Math": [
        80, 75, 90, np.nan, 85,
        78, 88, 92, 70, 95
    ],  # 10% missing

    "Salary": [
        30000, np.nan, np.nan, 40000, np.nan,
        35000, np.nan, 50000, np.nan, 38000
    ]  # 50% missing 
})
```

```python
#fill categorical column by mode 
#fill numerical colum by mean useing SimpleImputer()
df_numeric = df.select_dtypes(include=["int64", "float64"]).columns
df_categorical = df.select_dtypes(include=["str"]).columns
```

```python
#impute means filling 
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy ="mean") #medain, most_frequent
imputerValue = imputer.fit_transform(df[df_numeric])#it imput each column by its impute
print(imputerValue)
df[df_numeric] =imputerValue
df


```

```python
#impute categorical 
catImputer = SimpleImputer(strategy="most_frequent")
catImputerValue = catImputer.fit_transform(df[df_categorical])
df[df_categorical] = catImputerValue
df
```
