```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ram", "Shyam", "Hari", "Sita", "Gita"],
    "Age": [20, np.nan, 22, None, 21],
    "Math": [80, 75, np.nan, 90, None],
    "Science": [85, np.nan, 78, 92, 88],
    "City":["Kathmandu","Pokhara","Lalitpur",None,"Lalitpur"]
})

print(df)
```

```python
#removing rows containing null
# df.dropna(axis,inplace=True) #it will remove rows from original array
#axios=0 means row
#axios=1 means column
#dropna remove the row or column which have null value
df.dropna(axis=1) #it will drop column
```

```python
#thresh parameter
df.dropna(axis=1,thresh=4)
#remove those column 
# which has less than 
# 4 non null data
```

```python
df.dropna(axis=0) #it will remove rows
```

```python
df.dropna(axis=0,thresh=4)
```

```python
df.dropna(axis=0,thresh=100) 
#it does not show row as non of the row has at least 100 non null data
```
