```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Ram", "Shyam", "Hari", "Sita", "Gita"],
    "Age": [20, np.nan, 22, None, 21],
    "Math": [80, 75, np.nan, 90, None],
    "Science": [85, np.nan, 78, 92, 88],
    "City":["Kathmandu","Pokhara","Lalitpur",None,"Lalitpur"]
})

print(df)
```

```python
#find null table 
#it gives the table which have nan value
df.isna()
```

```python
#it gives wether a column has null value
df.isna().any()
```

```python
#finding total number of null value in each column
df.isna().sum()
```

```python
#finding total number of null value

df.isna().sum().sum()
```
