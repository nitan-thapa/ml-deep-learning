```python
import pandas as pd

data = {
    "Name": ["Ram", "Sita", "Hari", "Gita", "Nabin", "Asha", "Rohan", "Mina"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Math": [85.2134, 90.67812, 70, 95, 60, 88, 75, 92],
    "Science": [80, 85, 75, 90, 65, 95, 70, 89],
    "City": ["Kathmandu", "Pokhara", "Kathmandu", "Lalitpur",
             "Bhaktapur", "Pokhara", "Kathmandu", "Lalitpur"]
}

df = pd.DataFrame(data)
```

```python
#adding English columns at last
df["English"]=[80,85,70,95,60,88,75,92] #add #english column to last
df
```

```python
#Nottttttttttttt impppppppppppp
#addin columns using insert method
#here we can control the position of the new column unlinke direct method 
#df.insert(index_position, column_name, values
df.insert(3,"Nepali", [80,85,70,95,60,88,75,92])# 3 is the index

df
```

```python
#changing whole column
df["Math"] = df["Math"] + 5
df
```

```python
df["Math"]=df["Math"].round(1) #it is imp
df
```
