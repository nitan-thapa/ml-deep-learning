#Saving Dataframe


```python
import pandas as pd

data = {
    "Name":["Ram","Shyam"],
    "Age":[28,29],
    "salary":[50000,100000]
}

df= pd.DataFrame(data)

```

```python
df.to_csv("data.csv", index=False)#index=False means we don't want to save the index column
df.to_excel("data.xlsx",index=False)#you need to install openpyxl if needed
df.to_json("data.json",index=False)#you need to install openpyxl if neededs
```
