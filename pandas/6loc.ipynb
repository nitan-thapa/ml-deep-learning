.loc[] is used to  get

row  by index 

column by Name


df.loc[row_index]

df.loc[row_index, column_name]

for single row_index use 1 

for multiple row_index  use [1,2]

for single column_name use "column_name"

for multiple column_name use ["column_name1","column_name2"]




```python
import pandas as pd

data = {
    "Name": ["Ram", "Sita", "Hari", "Gita", "Nabin", "Asha", "Rohan", "Mina"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Math": [85, 90, 70, 95, 60, 88, 75, 92],
    "Science": [80, 85, 75, 90, 65, 95, 70, 89],
    "City": ["Kathmandu", "Pokhara", "Kathmandu", "Lalitpur",
             "Bhaktapur", "Pokhara", "Kathmandu", "Lalitpur"]
}

df = pd.DataFrame(data)
df
```

```python

# df.loc[row_index, column_name]=new_value
#for single row_indes use 1 
#for multiple row_index  use [1,2]
#for single column_name use "column_name"
#for multiple column_name use ["column_name1","column_name2"]




```

```python
#selectin a row by index level
print(df.loc[0])
```

```python
#selecting a multiple rows by index level
print(df.loc[[0,1,2,3]])
```

```python
#selectin a multiple rows using slicing
print(df.loc[:3])#it include end index as well
```

```python
#selectin row and column
df.loc[[0,1],"Math"]#select 0 and 1 row and give Math column
```

```python
df.loc[[0,1],["Math","Science"]]#show  0 and 1 row and show Math and Science column
```

```python
df.loc[:,["Math","Science"]]#show all row and show Math and Science column
```

```python
df
```

```python
df.loc[:,"Name":"Math"]#show all row and show Name to Math
#it include Math as well
```

```python

print(df["Math"][0])#we can get specific cell
df["Math"][0]=90  #it can not update specific cell
df
```

```python
print(df.loc[0,"Math"])#we can get specific cell
df.loc[0,"Math"]=90  #it can  update specific cell
df
```
