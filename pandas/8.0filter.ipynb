##filter row by column condition


```python
import pandas as pd

data = {
    "Name": ["Ram", "Sita", "Hari", "Gita", "Nabin", "Asha", "Rohan", "Mina"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Math": [85, 90, 70, 95, 60, 88, 75, 92],
    "Science": [80, 85, 75, 90, 65, 95, 70, 89],
    "City": ["Kathmandu", "Pokhara", "Kathmandu", "Lalitpur",
             "Bhaktapur", "Pokhara", "Kathmandu", "Lalitpur"]
}

df = pd.DataFrame(data)
df
```

```python
df[df["City"] == "Pokhara"]
```

```python
df[(df["Math"] > 80) & (df["Science"] > 80)]  #must give ()
```

```python
df[(df["City"] == "Pokhara")|(df["City"] == "Lalitpur"  )]
```

```python
df[~(df["City"]=="Pokhara")]
```

```python
#using loc
df.loc[df["Math"]>80]  #ge
```

```python
df.loc[:,df.columns != "City"]  #you can select columns using filter condition
```
