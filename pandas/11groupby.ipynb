```python
import pandas as pd
data = {
    "Name": ['nitan', 'sita', 'gita',"ram"],
    "Gender": ["male","female","female","male"],
    "Salary": [10000, 20000, 30000,10000]
}
df=pd.DataFrame(data)
```

```python
gender_salary=df.groupby("Gender")["Salary"].sum()
#Male
    # name gender  salary
    # nitan male 1000
    # ram   male 1000

#Female
    # name gender  salary
    # sita  female 2000
    # gita   female 3000
print(gender_salary)
print(gender_salary["female"])
```

```python
#transform (it will create new column for each row of gender and it will calculate mean  salary for each gender)
df.groupby("Gender")["Salary"].transform("mean")
# male 
# "Name" :["nitan","ram"]
# "Salary":[10000,1000]

# female
# "Name" :["sita","gita"]
# "Salary":[20000,30000]
```
