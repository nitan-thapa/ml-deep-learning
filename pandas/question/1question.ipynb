```python
# create average column of  math, scienc, english 
#create result if  marks>=80 then distinction  if marks>=60 then first else pass
# df["Average"] = (df["Math"]+df["Science"]+df["English"])/3
# df

#whye NaN +1 = NaN thus before averagint you need to replace NaN value 

df.fillna({
    "Math":df["Math"].mean(),
    "Science":df["Science"].mean(),
    "English":df["English"].mean()
    },inplace=True)
df
df["Average"] = (df["Math"]+df["Science"]+df["English"])/3  
df["Result"] = [ "Distinction" if el>=80 else "First" if el>=60 else "Pass" for el in df["Average"]]
df



```
