```python
import pandas as pd
data1 = {
    "Name": ['nitan', 'ram', 'hari',"gita"],
    "Age": [28, 34, 22,34],
    "Salary": [10000, 20000, 30000,10000]
}
data2 = {
    "Name": ["ram","hari"],
    "Age": [28, 34],
    "Salary": [10000, 20000]
}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
```

```python
df  = pd.concat([df1,df2],axis=0,ignore_index=True)
#axis=0 means ro2
#axis = 1 means column
df
```
