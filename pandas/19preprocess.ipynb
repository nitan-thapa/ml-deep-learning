```python
# shape
# info
# describe
# head
# tail



import pandas as pd
df = pd.DataFrame({
    "Study Hours": [2, 3, 4, 5, 6, 7, 1, 8],
    "Attendance": [60, 65, 70, 75, 80, 85, 50, 90],
    "Gender": ["Male", "Female", "Male", "Female",
               "Male", "Female", "Male", "Female"],
    "Result": ["Fail", "Fail", "Pass", "Pass",
               "Pass", "Pass", "Fail", "Pass"]
})
print(df)
```

```python
# shape
print(df.shape ) # (8, 4)  (8 rows, 4 col)
print(df.shape[0])
print(df.shape[1])
```

```python
df.info()
```

```python
print(df.describe())
print(df.describe()["Study Hours"]["25%"])
print(df.describe()["Study Hours"]["75%"])
```
