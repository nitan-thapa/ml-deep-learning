```python
#barplot (2 variable plot)

# it is a plot between categorical variable and numberic variable using bar


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "course": ["Python", "Java", "MERN"],
    "fee": [10000, 12000, 15000]
})
```

```python
sns.barplot(x=df["course"], y=df["fee"])
plt.show()
```
