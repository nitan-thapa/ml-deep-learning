```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset("tips")
print(df.head())
df.info()
```

```python
#make a countplot for all categorical variabel
#make a histplot for all numeric variable
```

```python
df_categorical = df.select_dtypes(include=["category"])
df_categorical
```

```python
plt.figure(figsize=(10,5))
for i, value in enumerate(df_categorical.columns):#[0,1,2,3]
    plt.subplot(2,2,i+1)
    sns.countplot(x=df[value])
    
    plt.tight_layout()
```

```python
df_numeric = df.select_dtypes(include=["int64","float64"])
df_numeric
```

```python
plt.figure(figsize=(10,5))
for i , value in enumerate(df_numeric.columns):#[0,1,2,3]
    plt.subplot(2,2,i+1)
    sns.histplot(df[value])
    
    plt.tight_layout()
```
