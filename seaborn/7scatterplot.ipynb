```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.DataFrame({
    "age": [20, 22, 25, 28, 30, 32, 35, 38, 40, 45],
    "income": [20000, 25000, 30000, 35000, 40000,
               45000, 50000, 55000, 60000, 70000],
    "purchased": ["No", "No", "No", "Yes", "No",
                  "Yes", "Yes", "Yes", "Yes", "Yes"]
})

```

```python
sns.scatterplot(
    x=df["age"],
    y=df["income"],
    hue=df["purchased"] #it is used to seperate the points based on hue
)


plt.xlabel("Age")
plt.ylabel("Income")
plt.show()
```
